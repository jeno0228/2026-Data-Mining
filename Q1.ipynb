```python
lines = sc.textFile("sherlock.txt") # 1줄을 원소 하나로 하는 RDD
lines.take(10) # take(5); 앞 5개만 보여주는 action
```

```python
words = lines.flatMap(lambda line: line.split(" ")) # 공백 기준으로 쪼갠 list 돌려줌
words.take(10)
#*** map말고 flatMap을 쓰는 이유: 납작하게 펼쳐 하나의 list로 만들기 위해.
```

```python
# 길이가 2보다 큰 단어만 고려하기
long_words = words.filter(lambda x: len(x)>=2)
long_words.take(10)
```

```python
# 각 단어의 앞 2글자만 빼오기 (key, 1)
two_spell = long_words.map(lambda x: (x[:2].lower(), 1))  # lower: 대소문자 구분X하게.
two_spell.take(10)
```

```python
# 2글자 빈도 세기
counts = two_spell.reduceByKey(lambda a,b: a+b)
counts.take(5)
```

```python
#빈도 32 미만인 2글자들 빈도 합
small = counts.filter(lambda x: x[1]< 32)
counts_sum = small.map(lambda x: x[1]).sum()
print(counts_sum)
```
