```python
import wget
wget.download("https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt", "int-pair-input.txt")
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local").getOrCreate()
sc = spark.sparkContext
```

```python
input_data = sc.textFile("int-pair-input.txt")
```

```python
# 문제가 전부 정수 쌍 중 첫번째 원소만 묻기 때문에 따로 추출한다
firsts = input_data.map(lambda line: int(line.split(",")[0][1:]))
print(firsts.collect())
```

```python
# 첫번째 정수 쌍의 첫번쨰 원소 (a)
a = firsts.first()
print(a)
```

```python
# 모든 정수 쌍의 첫번째 원소중 최댓값 (b)
b = firsts.reduce(lambda x,y: x if x > y else y)
print(b)
```

```python
# 모든 정수 쌍의 첫번째 원소중 최솟값 (c)
c = firsts.reduce(lambda x,y: x if x < y else y)
print(c)
```

```python
# 첫번째 원소가 1인 정수 쌍의 개수 (d)
d = firsts.filter(lambda x: x==1).count()
print(d)
```

```python
# 첫 번째 원소가 2인 정수 쌍의 개수 (e)
e = firsts.filter(lambda x: x==2).count()
print(e)
```

```python
# 합계
print(a+b+c+d+e)
```
